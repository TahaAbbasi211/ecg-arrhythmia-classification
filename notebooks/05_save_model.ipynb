```python
import sys
sys.path.append("..")

import os
import numpy as np
import joblib
from sklearn.ensemble import RandomForestClassifier
from sklearn.utils.class_weight import compute_sample_weight
from src.features import build_features

d = np.load("../data/processed/beats.npz")
m = d["y_train"] != 4
F = build_features(d["X_train"][m], d["rr_train"][m], d["g_train"][m])
y = d["y_train"][m]

model = RandomForestClassifier(n_estimators=200, min_samples_leaf=2,
                               n_jobs=-1, random_state=42)
model.fit(F, y, sample_weight=compute_sample_weight("balanced", y))

os.makedirs("../models", exist_ok=True)
joblib.dump(
    {"model": model, "weights": np.array([1.0, 4.0, 1.0, 8.0]),
     "classes": ["N", "S", "V", "F"]},
    "../models/ecg_rf.joblib", compress=3)

print("حجم فایل:", round(os.path.getsize("../models/ecg_rf.joblib") / 1e6, 1), "MB")
```
